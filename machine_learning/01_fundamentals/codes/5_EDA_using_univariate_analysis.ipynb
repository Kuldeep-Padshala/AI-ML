Univariate analysis: 
Independent analysis of each column.
 1. Numerical columns: distribution, outliers, skewness, kurtosis.
 2. Categorical columns: frequency distribution, mode, unique values.
Univariate analysis helps to understand the individual characteristics of each feature, identify potential issues, and inform subsequent analysis steps.

```python
df = pd.read_csv('/content/drive/MyDrive/datasets/train.csv')
```

```python
df.head()
```

## 1. Categorical Data

```python
# categorical features
df[['Survived', 'Pclass', 'Sex', 'SibSp', 'Parch', 'Embarked']].head()

# Survived -> values 0 and 1 (0 -> not survived, 1 -> survived)
# Pclass -> values 1, 2 and 3 (classes)
# Sex -> values male and female
# SibSp -> values 0, 1, 2, 3, 4, 5 and 8 (siblings/spouse aboarded)
# Parch -> values 0, 1, 2, 3, 4, 5 and 6 (parents/children aboarded)
# Embarked -> values C, Q and S (C -> Cherbourg, Q -> Queenstown, S -> Southampton)
```

### a. Countplot

```python
sns.countplot(data=df, x='Survived', palette='Set2')  # type: ignore
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(data=df, x='Pclass', palette='Set2')  # type: ignore
```

### b. PieChart

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

## 2. Numerical Data

### a. Histogram

```python
import matplotlib.pyplot as plt     # type: ignore
plt.hist(df['Age'], bins=10, edgecolor='black')
```

### b. Distplot

```python
sns.displot(df['Age'], kde=True)
```

### c. Boxplot

```python
# horizontal box plot
sns.boxplot(data=df, x='Age', orient='h')
```
