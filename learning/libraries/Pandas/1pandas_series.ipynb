### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np, pandas as pd
```

### Series from lists

```python
# string
country = ['India','Pakistan','USA','Nepal','Srilanka']

pd.Series(country)
```

```python
# integers
runs = [13,24,56,78,100]

runs_ser = pd.Series(runs)
runs_ser
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

```python
# custom index using dictionary

dic = {'maths':67,'english':57,'science':89,'hindi':100}
pd.Series(dic)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='My marks')
marks
```

### Series from dict

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}

marks_series = pd.Series(marks,name='My marks')
marks_series
```

```python
temp = pd.Series([1,2,3,np.nan,5,6,np.nan,8,np.nan,10])
temp
```

```python
temp.count()   # it will count the number of non-null values in the series
```

```python
# dropna
temp.dropna()    # it will drop the null values from the series
```

```python
# fillna
temp.fillna(temp.mean())   # it will fill the null values with the mean of the series
```
